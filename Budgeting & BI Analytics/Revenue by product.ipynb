```python
sales_data = {
    "Product": ["Nameplates", "Frames", "Engraved Items"],
    "Units Sold": [2000, 1500, 1000],
    "Price per Unit": [200, 150, 300],
    "Revenue": [400000, 225000, 300000]
}
sales_df = pd.DataFrame(sales_data)
sales_df
```

```python
sales_df["% of Total"] = ((sales_df["Revenue"] / sales_df["Revenue"].sum()) * 100).round(2).astype(str) + "%"
sales_df
```

```python
import matplotlib.pyplot as plt

plt.bar(sales_df["Product"], sales_df["Revenue"], color="steelblue")
plt.title("Revenue by Product (FY2026 Budget)")
plt.xlabel("Product")
plt.ylabel("Revenue (INR)")
plt.show()
```
